# Seasonal Agriculture Performance Analysis

## VOIS AICTE Data Analytics Major Project

### Objective

The objective of this project is to analyze agricultural data from different seasons and identify meaningful patterns, trends, relationships and differences in agricultural performance.

This project includes data cleaning, exploratory data analysis, visualization, seasonal comparison, environmental analysis, resource usage analysis and economic performance analysis.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())

df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(df["Soil_Moisture_pct"].median())

df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(df["Yield_Tonnes_Ha"].median())

print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
print("Number of Duplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", df.shape)
```

```python
df.describe()
```

```python
print(df["Season"].value_counts())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Season")

plt.title("Number of Agricultural Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")

plt.show()
```

```python
season_performance = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Revenue_INR": "mean",
    "Profit_INR": "mean",
    "Rainfall_mm": "mean",
    "Water_Used_m3": "mean",
    "Water_Efficiency_t_per_1000m3": "mean"
}).round(2)

season_performance
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Yield_Tonnes_Ha",
    estimator="mean"
)

plt.title("Average Agricultural Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes per Hectare)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Profit_INR",
    estimator="mean"
)

plt.title("Average Agricultural Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")

plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Rainfall_mm",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.7
)

plt.title("Relationship Between Rainfall and Agricultural Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes per Hectare)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Rainfall_mm",
    estimator="mean"
)

plt.title("Average Rainfall by Season")
plt.xlabel("Season")
plt.ylabel("Average Rainfall (mm)")

plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Avg_Temperature_C",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.7
)

plt.title("Relationship Between Temperature and Agricultural Yield")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Yield (Tonnes per Hectare)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Water_Used_m3",
    estimator="mean"
)

plt.title("Average Water Usage by Season")
plt.xlabel("Season")
plt.ylabel("Average Water Used (m³)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Water_Efficiency_t_per_1000m3",
    estimator="mean"
)

plt.title("Average Water Efficiency by Season")
plt.xlabel("Season")
plt.ylabel("Water Efficiency")

plt.show()
```

```python
crop_yield = (
    df.groupby("Crop")["Yield_Tonnes_Ha"]
    .mean()
    .sort_values(ascending=False)
)

print(crop_yield)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=crop_yield.index,
    y=crop_yield.values
)

plt.title("Average Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes per Hectare)")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(11, 6))

sns.barplot(
    data=df,
    x="Crop",
    y="Yield_Tonnes_Ha",
    hue="Season",
    estimator="mean"
)

plt.title("Crop Yield Performance Across Seasons")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes per Hectare)")
plt.xticks(rotation=45)

plt.legend(title="Season")

plt.show()
```

```python
state_yield = (
    df.groupby("State")["Yield_Tonnes_Ha"]
    .mean()
    .sort_values(ascending=False)
)

print(state_yield)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=state_yield.index,
    y=state_yield.values
)

plt.title("Average Agricultural Yield by State")
plt.xlabel("State")
plt.ylabel("Average Yield (Tonnes per Hectare)")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Revenue_INR",
    estimator="mean"
)

plt.title("Average Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Average Revenue (INR)")

plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Revenue_INR",
    y="Profit_INR",
    hue="Season",
    alpha=0.7
)

plt.title("Relationship Between Revenue and Profit")
plt.xlabel("Revenue (INR)")
plt.ylabel("Profit (INR)")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Agricultural Variables")

plt.show()
```

```python
yield_correlation = (
    numeric_df.corr()["Yield_Tonnes_Ha"]
    .sort_values(ascending=False)
)

print(yield_correlation)
```

```python
season_summary = df.groupby("Season").agg(
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Production=("Production_Tonnes", "mean"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Rainfall=("Rainfall_mm", "mean"),
    Average_Water_Used=("Water_Used_m3", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean")
).round(2)

season_summary
```

# Key Findings

1. Agricultural performance varies significantly across different seasons.

2. Kharif season shows comparatively strong agricultural performance in terms of yield, production and profitability.

3. Seasonal rainfall patterns show noticeable variation and may influence agricultural productivity.

4. Water usage and water efficiency differ across seasons, indicating differences in resource requirements.

5. Crop performance is not uniform across seasons, suggesting that some crops perform better under particular seasonal conditions.

6. Economic outcomes such as revenue and profit also vary between seasons.

7. The correlation analysis identifies relationships between agricultural yield and other environmental, production and resource-related variables.

8. State-wise analysis shows that agricultural performance can vary geographically as well as seasonally.

# Recommendations

1. Seasonal agricultural planning should consider differences in rainfall, temperature and soil conditions.

2. Crops should be selected based on their performance and suitability during different seasons.

3. Farmers and agricultural planners should focus on improving water efficiency, especially during seasons with lower rainfall.

4. Resource usage such as irrigation, fertilizers and other farming inputs should be optimized based on seasonal requirements.

5. Regions showing comparatively lower yield or profitability should be further investigated to identify the underlying causes.

6. Agricultural planning should use historical seasonal data to support data-driven decision making.

7. Regular monitoring of environmental and economic factors can help improve agricultural productivity and profitability.

# Conclusion

This project analyzed seasonal agricultural performance using data analytics techniques.

The analysis showed that agricultural performance varies across seasons in terms of yield, production, rainfall, water usage, revenue and profitability. Seasonal environmental conditions and resource usage patterns appear to be associated with differences in agricultural outcomes.

The analysis also identified variations in crop performance and state-wise agricultural productivity. Correlation analysis helped explore relationships between agricultural yield and other numerical variables.

Overall, the findings demonstrate that data analytics can be used to identify meaningful seasonal patterns and support better agricultural planning. The results can help farmers, agricultural planners and other stakeholders make more informed decisions regarding crop selection, resource allocation and seasonal farming strategies.
